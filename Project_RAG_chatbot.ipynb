```python
faq_text_data = """
Question: How can I reset my password?
Answer: Go to Settings, select Security, and click Reset Password.
You will receive instructions to create a new password.

Question: How long does shipping take?
Answer: Standard shipping usually takes 3 to 5 business days.
Express shipping usually takes 1 to 2 business days.

Question: Can I get a refund?
Answer: You can request a refund within 30 days of purchase.
The product must be eligible for a refund.

Question: How do I track my order?
Answer: Open My Orders and select the order you want to track.
You will see the current delivery status and tracking information.

Question: How can I change my email address?
Answer: Go to Account Settings and select Email Address.
Enter your new email address and verify it.
"""
```

```python
from langchain_text_splitters import TokenTextSplitter

text_splitter = TokenTextSplitter(
    chunk_size=100,
    chunk_overlap=20
)

chunks = text_splitter.split_text(faq_text_data)

print("Number of chunks:", len(chunks))
```

```python
for i, chunk in enumerate(chunks):
    print(f"\n--- Chunk {i + 1} ---")
    print(chunk)
```
